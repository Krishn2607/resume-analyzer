```python
import re

resume_text = """
John   Doe


Software   Engineer

Skills:    Python,   Java,   SQL,   Git


Experience:   2 years   at TCS


Education:   B.Tech   Computer Science   2023
"""

print("BEFORE CLEANING:")
print(resume_text)
```

```python
def clean_text(text):
    text = re.sub(r' +', ' ', text)
    text = re.sub(r'\n+', '\n', text)
    text = text.strip()
    return text

cleaned = clean_text(resume_text)
print("AFTER CLEANING:")
print(cleaned)
```
